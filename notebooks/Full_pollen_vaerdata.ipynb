```python


import pandas as pd

# 1. Load the historical data using semicolon separator
pollen = pd.read_csv("Zurich pollen 2023-2025.csv", sep=";")

# 2. Convert dates matching your exact pattern (using 'wet' instead of 'df')
pollen["reference_timestamp"] = pd.to_datetime(pollen["reference_timestamp"], format="%d.%m.%Y %H:%M")

# 3. Filter to keep only data from 2023 onwards
filtered_df = pollen[pollen["reference_timestamp"] >= "2023-01-01"]

# 4. Load the 2026 data (making sure to use semicolon separator as well)
pollen_2026 = pd.read_csv("Zurich pollen 2026.csv", sep=";")

# 5. Convert 2026 dates to the same format so they match perfectly
pollen_2026["reference_timestamp"] = pd.to_datetime(pollen_2026["reference_timestamp"], format="%d.%m.%Y %H:%M")

# 6. Vertically combine (stack) the filtered data and the 2026 data
full_pollen = pd.concat([filtered_df, pollen_2026], ignore_index=True)

# 7. Save the final combined dataset using your original semicolon format
full_pollen.to_csv("full_pollen_2023_2026.csv", sep=";", index=False)

# View the result
print(full_pollen)
```

```python

import pandas as pd

# 1. Load the historical data using semicolon separator
wet = pd.read_csv("Zurich weather data 2020-2025.csv", sep=";")

# 2. Convert dates matching your exact pattern (using 'wet' instead of 'df')
wet["reference_timestamp"] = pd.to_datetime(wet["reference_timestamp"], format="%d.%m.%Y %H:%M")

# 3. Filter to keep only data from 2023 onwards
filtered_df = wet[wet["reference_timestamp"] >= "2023-01-01"]

# 4. Load the 2026 data (making sure to use semicolon separator as well)
wet_2026 = pd.read_csv("Zurich weather data 2026.csv", sep=";")

# 5. Convert 2026 dates to the same format so they match perfectly
wet_2026["reference_timestamp"] = pd.to_datetime(wet_2026["reference_timestamp"], format="%d.%m.%Y %H:%M")

# 6. Vertically combine (stack) the filtered data and the 2026 data
full_weather = pd.concat([filtered_df, wet_2026], ignore_index=True)

# 7. Save the final combined dataset using your original semicolon format
full_weather.to_csv("full_weather_2023_2026.csv", sep=";", index=False)

# View the result
print(full_weather)

```

```python
import pandas as pd

# 1. Opprett master-tidslinjen og sett den som indeks
master_time = pd.date_range(start="2023-01-01 00:00", end="2026-09-09 23:00", freq="H")
df_master = pd.DataFrame(index=master_time)
df_master.index.name = "master_timestamp"

# 2. Last inn settene
d1 = pd.read_csv("full_pollen_2023_2026.csv", sep=";")
d2 = pd.read_csv("full_weather_2023_2026.csv", sep=";")

# 3. Konverter tidskolonner til datetime-objekter
d1["reference_timestamp"] = pd.to_datetime(d1["reference_timestamp"], format="mixed")
d2["reference_timestamp"] = pd.to_datetime(d2["reference_timestamp"], format="mixed")

# 4. Lag en kopi av tidsstemplene med unike navn FØR merge så de ikke forsvinner
d1["timestamp_pollen"] = d1["reference_timestamp"]
d2["timestamp_weather"] = d2["reference_timestamp"]

# 5. Sett tidsstemplene som indeks på kildedataene
d1 = d1.set_index("reference_timestamp")
d2 = d2.set_index("reference_timestamp")

# 6. Merge begge tabellene inn i master-tidslinjen basert på indeksen
# Vi bruker suffikser i tilfelle tabellene har andre like kolonnenavn (f.eks. 'verdi', 'sted')
synced_df = df_master.join(d1, how="left")
synced_df = synced_df.join(d2, how="left", lsuffix="_pollen", rsuffix="_weather")

# 7. Flytt master-tidslinjen tilbake fra indeks til en vanlig kolonne
synced_df = synced_df.reset_index()

# Se resultatet (Sjekk kolonnene 'timestamp_pollen' og 'timestamp_weather')
print(synced_df.head(20))

# Lagre den synkroniserte master-filen
synced_df.to_csv("full_pollen_and_weather_data_Zurich.csv", sep=";", index=False)

```
